```python
# Riley Balke
# gpa_checker.py
# Accepts student names and GPAs from user input. (First and last names.)
# It checks whether each student qualifies for the Dean's List 
# (GPA >= 3.5) or the Honor Roll (GPA >= 3.25) and displays the result.
# The program runs continuously until 'ZZZ' is entered as a last name.
# https://github.com/RileyBalkeIvyTech/Module-2-Lab---Case-Study-if...else-and-while

def main():
    print("--- Student GPA Qualification Checker ---")
    print("Enter 'ZZZ' at the last name prompt to quit.\n")

    while True:
        last_name = input("Enter student's last name: ").strip()

        if last_name.upper() == 'ZZZ':
            print("Exiting program. Goodbye!")
            break

        first_name = input("Enter student's first name: ").strip()

        try:
            gpa = float(input(f"Enter GPA for {first_name} {last_name}: "))
        except ValueError:
            print("Invalid input. Please enter a valid decimal number for the GPA.")
            continue

        full_name = f"{first_name} {last_name}"

        if gpa >= 3.5:
            print(f" {full_name} has made the Dean's List!")
        elif gpa >= 3.25:
            print(f" {full_name} has made the Honor Roll!")
        else:
            print(f"Student {full_name} does not qualify for the Honor Roll or Dean's List.")
        
        print()

if __name__ == "__main__":
    main()
```
